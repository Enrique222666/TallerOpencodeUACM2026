# Comandos Slash de OpenCode

Guía rápida de los comandos slash disponibles en el TUI de OpenCode.

---

## 📋 Lista de Comandos

| Comando | Alias | Atajo | Descripción |
|---------|-------|-------|-------------|
| `/connect` | - | - | Agrega un proveedor de modelos y configura su API key |
| `/models` | - | `Ctrl+X M` | Abre el selector de modelos disponibles |
| `/timeline` | - | `Ctrl+X G` | Muestra el historial y línea de tiempo de la sesión |
| `/themes` | - | `Ctrl+X T` | Lista y cambia entre los temas visuales disponibles |
| `/clear` | `/new` | `Ctrl+X N` | Inicia una nueva sesión (la anterior se guarda en el historial) |
| `/new` | `/clear` | `Ctrl+X N` | Inicia una nueva sesión |
| `/sessions` | `/resume`, `/continue` | `Ctrl+X L` | Lista y permite cambiar entre sesiones anteriores |
| `/compact` | `/summarize` | `Ctrl+X C` | Compacta (resume) la sesión actual para liberar contexto |
| `/share` | - | `Ctrl+X S` | Comparte la sesión actual como enlace público |

---

## 🔍 Detalle de cada comando

### `/connect`
Agrega un proveedor de modelos a OpenCode. Te permite seleccionar entre los proveedores disponibles (Anthropic, OpenAI, Google, etc.) y configurar sus API keys.
